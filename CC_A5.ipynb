### Q.1 Write a program to create a NumPy 1D-array with 5 elements and perform basic operations like:
a) Addition of 2 in all the elements
b) Multiply 3 with all the elements
c) Divide every element by 2

```python
# Create a NumPy 1D-array with 5 elements
arr = np.array([1, 2, 3, 4, 5])
print(f"Original array: {arr}")

# a) Addition of 2 to all elements
arr_add_2 = arr + 2
print(f"Array after adding 2: {arr_add_2}")

# b) Multiply 3 with all elements
arr_multiply_3 = arr * 3
print(f"Array after multiplying by 3: {arr_multiply_3}")

# c) Divide every element by 2
arr_divide_2 = arr / 2
print(f"Array after dividing by 2: {arr_divide_2}")
```

### Q.2 Questions on Basic NumPy Array:
a) Reverse the NumPy array: `arr = np.array([1, 2, 3, 6, 4, 5])`
b) Find the most frequent value and their indice(s) in the following arrays:
   i. `x = np.array([1,2,3,4,5,1,2,1,1,1])`
   ii. `y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])`

```python
# a) Reverse the NumPy array
arr_q2a = np.array([1, 2, 3, 6, 4, 5])
reversed_arr_q2a = arr_q2a[::-1]
print(f"Original array (Q2a): {arr_q2a}")
print(f"Reversed array (Q2a): {reversed_arr_q2a}")

# b) Find the most frequent value and their indice(s)
from collections import Counter

def most_frequent_value_and_indices(arr):
    counts = Counter(arr)
    if not counts: # Handle empty array case
        return None, []
    most_common = counts.most_common(1)
    most_frequent_val = most_common[0][0]
    indices = np.where(arr == most_frequent_val)[0]
    return most_frequent_val, indices

# i. Array x
x = np.array([1,2,3,4,5,1,2,1,1,1])
most_freq_x, indices_x = most_frequent_value_and_indices(x)
print(f"\nArray x: {x}")
print(f"Most frequent value in x: {most_freq_x}")
print(f"Indices of most frequent value in x: {indices_x}")

# ii. Array y
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])
most_freq_y, indices_y = most_frequent_value_and_indices(y)
print(f"\nArray y: {y}")
print(f"Most frequent value in y: {most_freq_y}")
print(f"Indices of most frequent value in y: {indices_y}")
```

### Q.4 Write program to create an 1-D NumPy array named BismanSinghRai with evenly spaced 25 numbers from 10 to 100 using `linspace()`. Print the dimensions of the array, shape, total elements, the data type of each element and total number of bytes consumed by the array. Find the transpose of this array using `reshape()` attribute. Can we do the same with `T` attribute?

```python
# Name: Bisman Singh Rai | Roll No: 1024160130
bisman_singh_rai_array = np.linspace(10, 100, 25)
print(f"1D NumPy array '{'BismanSinghRaiArray'}':\n{bisman_singh_rai_array}")

# Print dimensions
print(f"\nDimensions (ndim): {bisman_singh_rai_array.ndim}")

# Print shape
print(f"Shape: {bisman_singh_rai_array.shape}")

# Print total elements
print(f"Total elements (size): {bisman_singh_rai_array.size}")

# Print data type of each element
print(f"Data type of each element (dtype): {bisman_singh_rai_array.dtype}")

# Print total number of bytes consumed by the array
print(f"Total bytes consumed: {bisman_singh_rai_array.nbytes}")

# Find the transpose of this array using reshape() attribute
# For a 1D array, transpose doesn't change anything unless reshaped to 2D first.
# To see a meaningful 'transpose', we first need to reshape it into a 2D array (e.g., a column vector).

# Reshape to a column vector (25 rows, 1 column)
reshaped_col_vector = bisman_singh_rai_array.reshape(25, 1)
print(f"\nReshaped to column vector (25x1):\n{reshaped_col_vector}")
print(f"Shape of column vector: {reshaped_col_vector.shape}")

# Now, find the transpose of this 2D reshaped array using reshape
transposed_by_reshape = reshaped_col_vector.reshape(1, 25)
print(f"\nTranspose of column vector using reshape (1x25):\n{transposed_by_reshape}")
print(f"Shape of transposed by reshape: {transposed_by_reshape.shape}")

# Can we do the same with T attribute?
# The .T attribute is used for transposing arrays.
# For a 1D array, .T returns the array itself, as there's no second dimension to transpose with.
# However, if we apply it to the 2D reshaped array (column vector):
transposed_by_T = reshaped_col_vector.T
print(f"\nTranspose of column vector using .T attribute (1x25):\n{transposed_by_T}")
print(f"Shape of transposed by .T: {transposed_by_T.shape}")

# Conclusion: The .T attribute works directly on 2D arrays (like the reshaped column vector)
# to perform a proper transpose. For a pure 1D array, .T returns the original array,
# so to get a 'transposed' effect (e.g., making it a row/column vector),
# you generally need to reshape it to 2D first, then .T can be effectively used.
```

### Q5. Create a 2-D Array of three rows and four columns, named `ucs420_BismanSinghRai>` with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Resize the array to two rows and three columns and name it as `resized_ucs420_BismanSinghRai>`.
