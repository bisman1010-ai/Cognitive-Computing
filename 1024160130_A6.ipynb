## Q1. Sensor Readings Analysis

```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
print("Original temperatures (C):", temperature)

# a. Add 2C using vectorization
corrected_temp = temperature + 2
print("a. Corrected temperatures:", corrected_temp)

# b. Convert to Fahrenheit
temp_fahrenheit = (9/5) * temperature + 32
print("b. Temperatures in Fahrenheit:", temp_fahrenheit)

# c. Boolean indexing for > 32C
high_temp_mask = temperature > 32
high_temperatures = temperature[high_temp_mask]
print("c. Readings > 32 C:", high_temperatures)

# d. Count readings > 32C
high_temp_count = np.sum(high_temp_mask)
print("d. Count exceeding 32 C:", high_temp_count)
```

### e. Why Vectorization and Boolean Indexing are better than loops:

* **Vectorization:** Instead of running slow Python loops over each element, NumPy offloads operations to compiled C code. It processes data parallelly using SIMD instructions, which is much faster and cleaner.
* **Boolean Indexing:** Instead of writing complex `if-else` blocks inside loops to check conditions, we generate a simple boolean mask. NumPy applies this mask directly in memory to filter data instantly.

## Q2. Daily Steps Data Analysis

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

# a. Total steps
total_steps = np.sum(steps)
print("a. Total steps:", total_steps)

# b. Mean steps
mean_steps = np.mean(steps)
print("b. Mean steps:", round(mean_steps, 2))

# c. Max and Min steps
max_steps = np.max(steps)
min_steps = np.min(steps)
print(f"c. Max: {max_steps}, Min: {min_steps}")

# d. Daily totals (column sum)
daily_totals = np.sum(steps, axis=0)
print("d. Daily totals:", daily_totals)

# e. User totals (row sum)
user_totals = np.sum(steps, axis=1)
print("e. User totals:", user_totals)

# f. Max steps position
flat_max_index = np.argmax(steps)
max_position = np.unravel_index(flat_max_index, steps.shape)
print("f. Max steps index position:", max_position)
```

## Q4. Predictive Assistive System using Linear Regression (OLS)

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

# a. Shape and dimensions
print(f"a. Shape of X: {X.shape}, Dimensions: {X.ndim}")

# b. Transpose
X_transpose = X.T
print("b. Transpose of X:\n", X_transpose)

# c. Matrix multiplication
product = X_transpose @ X
print("c. Matrix Product X.T @ X:\n", product)

# d. Matrix inverse
inverse_product = np.linalg.inv(product)
print("d. Inverse:\n", inverse_product)

# e. Ordinary Least Squares (OLS) beta estimation
coefficients = inverse_product @ X_transpose @ y
print("e. Beta Coefficients:", coefficients)
```

### f. What the Coefficients mean:

These coefficients represent the impact of each feature on the final assistance score:
1. **Sleep Hours:** For every extra hour of sleep, the assistance score shifts by roughly `-0.16` (slightly reducing required assistance).
2. **Activity Level:** Each unit increase in daily activity adds about `0.14` to the predicted assistance score.
3. **Stress Level:** Stress has the highest impact here—each unit increase adds about `10.06` to the score.

```python
# g. Prediction for new user
new_user = np.array([5, 40, 7])
predicted_score = new_user @ coefficients
print("g. Predicted score:", round(predicted_score, 2))
```
